# Pratica
#### Desenvolva um programa que solicite ao usuário a temperatura atual. Caso a temperatura seja superior a 30ºC, exiba uma mensagem recomendando que ele se hidrate

```python
temperatura = int(input("Qual sua temperatura atual?"))

if temperatura >= 30:
    print("Beba agua")
```

#### Solicite um numero e diga se é par ou impar

```python
a = int(input("Digite um valor"))

if a % 2 == 0:
    print("Este numero é par")
else:
    print("Este numero é impar")
```

# Valores Booleanos - Pratica

```python
saldo = 497.59

trans = float(input("Digite o valor da transação: "))

saldo_suficiente = saldo - trans

if saldo_suficiente > saldo:
    print("Não foi possivel completar a transação, transação maior que o saldo disponivel!")
else:
    print("Transação realizada")
```
