## Problema 1: Desenvolvendo uma Calculadora básica
*   Exiba uma mensagem explicando o funcionamento da calculadora
*   Solicite ao usuario dois valores numericos
*   Calcule
    *   Soma, subtração, multiplicação, divisao, divisao inteira, resto da divisao e potencia
*   Exiba todos os resultados na tela

```python
# Calculadora

print("Bem vindo a Calculadora Digital, insira dois valores e receberá o resultado das 7 operaçoes aritimeticas basicas")
x1 = float(input("Digite o primeiro digito: "))
x2 = float(input("Digite o segundo numero: "))

soma = x1+x2
sub = x1-x2
multi = x1*x2
div = x1/x2
rdiv = x1%x2
divint = x1//x2
pot = x1**x2

print(f"Os seguintes resultados são\nSoma\t\t{soma:.2f}\nSubtração\t{sub:.2f}\nMultiplicação\t{multi:.2f}\nDivisão\t\t{div:.2f}\nResto div\t{rdiv:.2f}\nDiv inteira\t{divint:.2f}\nPotencia\t{pot:.2f}")
```

## Problema 2: Controle de Produção
*   Armazene a produção de cada dia em uma variável.
*   Calcule a produção total da semana
*   Calcule a produção media diaria
*   Determine quantas peças faltariam para atingir a meta semana de 7000 peças
*   Exiba todas as informaçoes na tela

```python
# Controle de Produção
segunda = 1240
terça = 1315
quarta = 1288
quinta = 1360
sexta = 1297

producao_total = segunda + terça + quarta + quinta + sexta
print(f"Foram produzidas {producao_total} durante a semana")

media = producao_total / 5
print(f"A media diaria foi de {media} peças")

meta = 7000
print(f"Para antingir a meta semanal, será preciso mais {meta-producao_total} peças")
```

### Problema 4
*   Temperadura media da maquina
*   exiba manualmente a temperatura maxima e minima registrada
*   Calcule a amplitude termica

*   Considerando que a temperatura ideal de operação é 73ºC
*   Exiba uma mensagem utilizando uma f-string, informando a temperatura media da maquina, a amplitude termica e a diferença em relação a temperatura ideal, apresentando os valores com uma casa decimal

```python
temp1 = 72.3
temp2 = 73.1
temp3 = 74.0
temp4 = 72.8
temp5 = 71.5
temp6 = 75.2
temp7 = 73.6
temp8 = 74.4

media = (temp1 + temp2 + temp3 + temp4 + temp5 + temp6 + temp7 + temp8) / 8
print(f"A temperatura media da maquina durante operação foi {media:.1f}ºC")

print(f"Temperatura maxima durante operação:    {temp6}\nTemperatura minima durante operação:   {temp5} ")

amplitude_termica = temp6 - temp5
print(f"A amplitude termica durante operação foi de {amplitude_termica:.1f}")

temp_ideal = 73.0
print(f"A difereça entre a temperatura ideal para a temperatura media foi de {temp_ideal - media:.1f}")
```
